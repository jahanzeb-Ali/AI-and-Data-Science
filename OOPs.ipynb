```python
class Student:
    def __init__(self,roll_num, name,gender, course ):
        self.roll_num = roll_num
        self.name = name
        self.gender = gender
        self.course = course


    def attending_class(self):
        print(f"{self.name} is attending class ")

    def exam(self):
        print(f"{self.roll_num} is passed exam ")
    
```

```python
s1 = Student("001","shahzeb","male","DS")
```

```python
s1.attending_class()
```
