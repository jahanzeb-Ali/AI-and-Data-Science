```python
with open("friends.txt","w") as f:
    serial = 1
    while True:
        friend_name = input("Enter friend name or q to exit: ")
        if friend_name != "q":
            f.write(f"{serial}- {friend_name} \n")
            serial += 1
        else:
            break
```

```python
with open("friends.txt",'r+') as f:
    n = len(f.readlines())
    content = f.read()
    serial = n + 1
    while True:
        friend_name = input("Enter friend name or q to exit: ")
        if friend_name != "q":
            f.write(f"{serial}- {friend_name} \n")
            serial += 1
        else:
            break
    f.seek(0)
    content = f.read()
    print(content)
```
