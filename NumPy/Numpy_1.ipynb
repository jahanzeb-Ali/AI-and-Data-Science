```python
flatted_arr =  arr2.flatten() 
# when you dont known the size of array. and want to make it  1 deminsion array 
# key difference in flatten and ravel is flatten makes copy of array and ravel only view the original array it does not 
# copy array
flatted_arr[3:8] = 777
flatted_arr
```

```python
arr2 #does not effect original array
```

```python
ravelled_arr = arr2.ravel()
# when you dont known the size of array. and want to make it  1 deminsion array 
# key difference in flatten and ravel is flatten makes copy of array and ravel only view the original array it does not 
# copy array

ravelled_arr[3:8] = 999
ravelled_arr
```

```python
arr2 # effected the original array
```

# Indexing and slicing 

```python
arr2d = np.arange(1,26).reshape(5,5)
arr2d
```

```python
arr2d[3,3] # got only one value this is called indexing 
```

```python
arr2d[:, 1:3] # got same dimension array ( e.g 2d )
```

```python
arr2d[2:4,1:4]
```

```python
arr2d[2:4,:]
```

```python
arr2d[:,-1:]
```

```python
arr2d[2:3,1:2]
```

```python
arr2d[::2,::2] # :2 is step value 
```

```python
arr2d[:, :]
```
