```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def annual_salary(self):
        print(f"Annual salary of {self.name } is {self.salary * 12} ")
```

```python
employee_1 = Employee("jahanzeb",20000)
```

```python
employee_1.annual_salary()
```

```python
class Manager(Employee):
    def __init__(self, name, salary,department):
        super().__init__(name,salary)
        self.department = department
    def annual_salary(self,bounce):
        print(f"Annual salary of {self.name } is {self.salary * 12 + bounce} ")
```

```python
m1 = Manager("jahanzeb", 13000, "Finance", 2000)
```

```python
m2 = Manager("Adil", 12000, "CS", 1000)
```

```python
m1.annual_salary(1000)
```
