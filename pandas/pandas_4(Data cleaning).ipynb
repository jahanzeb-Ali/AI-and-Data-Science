## Data Cleaning
    -Missing Data 

```python
import pandas as pd
```

```python
movies = pd.read_csv("movies.csv")
```

```python
movies.info()
```

```python
movies.isna().sum()
```

```python
#  if  record have all null values it will droped, how="all" is a condition and axis=0 refers to records (rows) and
# axis=1 refers to columns (vertical/ feature) 
movies.dropna(how="all", axis=0)
```

```python
movies.dropna(how="any", axis=0) 
```

```python
# if record have five features that have data inside than dont drop that record otherwise drop that
movies.dropna(axis=0, thresh=4)
```

```python
int(movies.shape[1]*0.7)
```

```python
movies.dropna(axis=0, thresh=int(movies.shape[1]*0.7))
```

```python
#Aghar Budget	Profitability	OpenProfit	Year mai se kise ke bhi value missing ho tu record drop kardo warna 
# record ko rahne do
movies.dropna(axis=0, subset=["Budget","Profitability","OpenProfit","Year"])
```
