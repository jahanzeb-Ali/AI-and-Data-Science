### Data Wrangling: Join, Combine, and Reshape

```python
import pandas as pd
```

```python
df1 = pd.read_excel("employee_data.xlsx",sheet_name=0)
df2 = pd.read_excel("employee_data.xlsx",sheet_name=1)
df3 = pd.read_excel("employee_data.xlsx",sheet_name=2)
```

```python
df1.head()
```

```python
df2.head()
```

```python
df3.head()
```

```python
pd.concat([df2,df1],axis=0)
```

```python
pd.concat([df2,df1],axis=1)
```

```python
# By default it overlapping 
pd.merge(df1,df3) # it will only section those rows(single column value ) that are same in both dataframe
```

```python
df3.columns = ['Name', 'Sex', 'age', 'salary']
```

```python
pd.merge(df1,df3)
```

```python
pd.merge(df1,df3, left_on="First Name", right_on="Name")
```

```python
pd.merge(df1,df3, left_on="First Name", right_on="Name").drop(["Name"])
```
