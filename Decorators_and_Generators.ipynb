# Decorators


```python
A decorator is a function that wraps another function to extend or modify its behavior,
without changing the original function's code.
```

```python
def my_decorator(func):
    def wrapper():
        # Perform some actions before calling func()
        print("Before function call")

        # Call the original function
        func()

        # Perform some actions after calling func()
        print("After function call")
    
    return wrapper

@my_decorator
def my_function():
    print("Inside my_function")
```

```python
def my_decorator(func):
    def wrapper(*arg,**kwarg):
        print("Before the function runs")
        func(*arg,**kwarg)
        print("After the function runs")
    return wrapper

```

```python
@my_decorator
def say_hello(nam, email):
    print("Hello!")
    print(nam,email)
say_hello("ali","a@gmail")
```

```python
def repeat(n):
    def decorator(func):
        def wrapper(*args, **kwargs):
            print("**kwargs",**kwargs)
            print("**kwargs",*args)
            for _ in range(n):                func(*args, **kwargs)
        return wrapper
    return decorator
```

```python
@repeat(3)
def greet():
    print("Hi!")

greet() 
```

# Generators

```python
Generators are an efficient way to create iterators in Python. 
They offer a memory-friendly approach to working with data sequences, 
especially when dealing with large or infinite datasets. 
Generators enable lazy evaluation, producing elements one at a time as needed.
Generator functions are defined like regular functions, but use the "yield" keyword instead of "return". 
```

```python
def countdown(n):
    while n > 0:
        yield n
        n -= 1

```

```python
for num in countdown(5):
    print(num) 
```

```python
def memoize(f):
    mem = {}
    
    def memoized_funtion(n):
        if n not in mem:
            mem[n] = f(n)
            
        return mem[n]
        
    return memoized_funtion
```

```python
# @memoize
# without memoize function 
def fib(n):
    if n <= 1:
        return n
    return fib(n - 1) + fib(n - 2)   
    
```

```python
%time fib(36)
```

```python
@memoize 
def fib(n):
    if n <= 1:
        return n
    return fib(n - 1) + fib(n - 2)   
```

```python
%time fib(36)
```
